# Early-Stage Anomaly Detection - Exploratory Data Analysis

This notebook provides an exploratory analysis of the Smart Manufacturing IoT-Cloud Monitoring Dataset.

## 1. Data Loading
We load the primary dataset consisting of sensor telemetry.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('data/smart_manufacturing_data.csv')
df['timestamp'] = pd.to_datetime(df['timestamp'])
df.head()
```

## 2. Feature Distributions
Let's look at the distribution of sensor metrics like vibration, pressure, and energy consumption.

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
sns.histplot(df['vibration'], ax=axes[0])
sns.histplot(df['pressure'], ax=axes[1])
sns.histplot(df['energy_consumption'], ax=axes[2])
plt.tight_layout()
plt.show()
```
