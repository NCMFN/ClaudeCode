# Exploratory Data Analysis: Synthetic 19th-Century Telegraph Fault Telemetry

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
sys.path.append('..')
from src.features import load_data, feature_engineering

# Set plotting style
sns.set_theme(style="whitegrid")
```

```python
df = load_data('../data/telegraph_faults.csv')
df.head()
```

```python
df.info()
```

```python
# Check missingness
df.isnull().sum() / len(df)
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='fault_class')
plt.title('Distribution of Fault Classes')
plt.xticks(rotation=45)
plt.show()
```

```python
df_eng = feature_engineering(df)
plt.figure(figsize=(10, 8))
sns.heatmap(df_eng.corr(numeric_only=True), annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation Matrix')
plt.show()
```
